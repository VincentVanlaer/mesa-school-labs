```python
import mesa_reader as mr
import matplotlib.pyplot as plt
```

**Plot 1: The evolution around the bump of the location of the base of the convection zone, the peak of burning and the mean molecular weight discontinuity as a function of mass and radius ordinate**

```python
plt.rc('font', weight='bold')
plt.rc('xtick.major', size=5, pad=10)
plt.rc('ytick.major', size=5, pad=10)
plt.rc('xtick.minor', size=2.5, pad=10)
plt.rc('ytick.minor', size=2.5, pad=10)

fig = plt.figure(figsize=(7.5,9))

ax1 = fig.add_axes([0.1,0.05,0.39,0.07])
ax1.minorticks_on()
ax1.tick_params(axis='both', which='major', labelsize=8,pad=5,direction='in',bottom=True, top=True, left=True, right=True)
ax1.tick_params(axis='both', which='minor', labelsize=4,pad=5,direction='in',bottom=True, top=True, left=True, right=True)
ax1.set_xlabel(r"${\rm age\;[Gyr]}$", fontsize=10)
ax1.set_ylabel(r"${\rm T_{eff}\;[K]}$", fontsize=10)
#ax1.set_xlim(11.9,12.14)
#ax1.set_ylim(4000,5000)

ax2 = fig.add_axes([0.6,0.05,0.39,0.07])
ax2.minorticks_on()
ax2.tick_params(axis='y',labelleft='off')
ax2.tick_params(axis='both', which='major', labelsize=8,pad=5,direction='in',bottom=True, top=True, left=True, right=True)
ax2.tick_params(axis='both', which='minor', labelsize=4,pad=5,direction='in',bottom=True, top=True, left=True, right=True)
ax2.set_xlabel(r"${\rm age\;[Gyr]}$", fontsize=10)
ax2.set_ylabel(r"${\rm log(L/L_{\odot})}$", fontsize=10)
#ax2.set_xlim(11.9,12.14)
#ax2.set_ylim(0.5,1.75)

ax3 = fig.add_axes([0.1,0.15,0.39,0.25])
ax3.minorticks_on()
ax3.tick_params(axis='y',labelleft='off')
ax3.tick_params(axis='both', which='major', labelsize=8,pad=5,direction='in',bottom=True, top=True, left=True, right=True)
ax3.tick_params(axis='both', which='minor', labelsize=4,pad=5,direction='in',bottom=True, top=True, left=True, right=True)
ax3.set_ylabel(r"${\rm mass\;[M_{\odot}]}$", fontsize=10)
#ax3.set_xlim(11.9,12.14)
#ax3.set_ylim(0.15,0.35)
#ax3.set_yticks([0.15,0.20,0.25,0.30,0.35])

ax4 = fig.add_axes([0.6,0.15,0.39,0.25])
ax4.minorticks_on()
ax4.tick_params(axis='y',labelleft='off')
ax4.tick_params(axis='both', which='major', labelsize=8,pad=5,direction='in',bottom=True, top=True, left=True, right=True)
ax4.tick_params(axis='both', which='minor', labelsize=4,pad=5,direction='in',bottom=True, top=True, left=True, right=True)
ax4.set_ylabel(r"${\rm radius\;[R_{\odot}]}$", fontsize=10)
#ax4.set_xlim(11.9,12.14)
#ax4.set_ylim(0,1.5)
#ax4.set_yticks([0,0.5,1,1.5])

h = mr.MesaData('/Users/planetos/MESA_work/MESA_2025/Day4_RGB/Minilab1_1p15/LOGS/history.data')
ax1.plot(h.star_age/(10**9), 10**h.log_Teff)
ax2.plot(h.star_age/(10**9), h.log_L)
ax3.scatter(h.star_age/(10**9), h.cz_bot_mass, s=1, color='steelblue')
ax3.scatter(h.star_age/(10**9), h.mass_max_eps_nuc, s=1, color='black')
ax3.plot(h.star_age/(10**9), h.mdisc, linestyle='dashed', color='steelblue')

ax4.scatter(h.star_age/(10**9), h.cz_bot_radius, s=1, color='steelblue', label='bcz')
ax4.scatter(h.star_age/(10**9), h.radius_max_eps_nuc, s=1, color='black', label=r'max($\epsilon_{nuc}$)')
ax4.plot(h.star_age/(10**9), h.rdisc, linestyle='dashed', color='steelblue', label=r'$\mu$-discontinuity')

ax4.legend(markerscale=2., scatterpoints=1, fontsize=8)

plt.savefig('lab1_fig1',dpi=300,bbox_inches='tight')
plt.show()
```

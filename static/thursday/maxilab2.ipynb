```python
# Install mesa-reader if needed
!pip install mesa-reader

# Import packages
import mesa_reader as mr
import matplotlib.pyplot as plt

# Upload file
from google.colab import files
print("Please upload your MESA history.data file")
uploaded = files.upload()

# Automatically get the uploaded filename
filename = list(uploaded.keys())[0]

# Load the history file
history_data = mr.MesaData(filename)

# Extract columns
age = history_data.data('star_age')
delta_Pg = history_data.data('delta_Pg')
center_he4 = history_data.data('center_he4')

# Filter condition: delta_Pg not zero and center_he4 < 0.9
valid_points = (delta_Pg != 0) & (center_he4 < 0.9)

# Plotting
fig, axs = plt.subplots(2, 1, figsize=(6, 8))

# Plot delta_Pg vs age
axs[0].plot(age[valid_points], delta_Pg[valid_points], label='Your Model')
axs[0].set_xlabel('Age')
axs[0].set_ylabel('ΔΠ')
axs[0].legend()

# Plot delta_Pg vs center_he4
axs[1].plot(center_he4[valid_points], delta_Pg[valid_points], label='Your Model')
axs[1].invert_xaxis()
axs[1].set_xlabel('Central He4')
axs[1].set_ylabel('ΔΠ')
axs[1].legend()

plt.tight_layout()
plt.savefig('delta_Pg_plots.pdf')
plt.show()
```

```python
# Install tomso if needed
!pip install tomso

import matplotlib.pyplot as plt
import numpy as np
from tomso import mesa

# Upload files
from google.colab import files
print("Please upload history.data file for the overshooting case")
hist_OV = list(files.upload().keys())[0]
hist_data_OV = mesa.load_history(hist_OV)

print("Please upload history.data file for the penetrative convection case")
hist_PC = list(files.upload().keys())[0]
hist_data_PC = mesa.load_history(hist_PC)

print("Please upload history.data file for the semiconvection case")
hist_SC = list(files.upload().keys())[0]
hist_data_SC = mesa.load_history(hist_SC)

print("Please upload history.data file for the max overshooting case")
# if you haven't run the max overshoot case, you can download the history file
# on the lab website
hist_maxOV = list(files.upload().keys())[0]
hist_data_maxOV = mesa.load_history(hist_maxOV)
```

```python
#load data from history files
delta_Pg_OV = hist_data_OV["delta_Pg"]
he_core_radius_OV = hist_data_OV["he_core_radius"]
Yc_OV = hist_data_OV["center_he4"]
star_age_OV = hist_data_OV["star_age"]

delta_Pg_PC = hist_data_PC["delta_Pg"]
he_core_radius_PC = hist_data_PC["he_core_radius"]
Yc_PC = hist_data_PC["center_he4"]
star_age_PC = hist_data_PC["star_age"]

delta_Pg_SC = hist_data_SC["delta_Pg"]
he_core_radius_SC = hist_data_SC["he_core_radius"]
Yc_SC = hist_data_SC["center_he4"]
star_age_SC = hist_data_SC["star_age"]

delta_Pg_maxOV = hist_data_maxOV["delta_Pg"]
he_core_radius_maxOV = hist_data_maxOV ["he_core_radius"]
Yc_maxOV = hist_data_maxOV ["center_he4"]
star_age_maxOV = hist_data_maxOV ["star_age"]
```

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2,2, figsize=(10,8))
ax1.set_title(r'Period spacing')
ax1.plot(star_age_OV/1e9,delta_Pg_OV, color='red',label='OV')
ax1.plot(star_age_PC/1e9,delta_Pg_PC, color='darkgreen',label='PC')
ax1.plot(star_age_maxOV /1e9,delta_Pg_maxOV , color='blue',label='MaxOV')
ax1.plot(star_age_SC /1e9,delta_Pg_SC , color='goldenrod',label='SC')
ax1.set_xlabel('Age (Gyr)')
ax1.set_ylabel(r'$\Delta \Pi$ (s)')
ax1.legend()

ax2.plot(Yc_OV,delta_Pg_OV, color='red')
ax2.plot(Yc_PC,delta_Pg_PC, color='darkgreen')
ax2.plot(Yc_maxOV,delta_Pg_maxOV, color='blue')
ax2.plot(Yc_SC ,delta_Pg_SC , color='goldenrod')
ax2.set_xlabel(r'$Y_c$')
ax2.set_ylabel(r'$\Delta \Pi$ (s)')
ax2.invert_xaxis()

ax3.set_title(r'He core radius')
ax3.plot(star_age_OV/1e9,he_core_radius_OV, color='red',label='OV')
ax3.plot(star_age_PC/1e9,he_core_radius_PC, color='darkgreen',label='PC')
ax3.plot(star_age_maxOV/1e9,he_core_radius_maxOV, color='blue',label='MaxOV')
ax3.plot(star_age_SC /1e9,he_core_radius_SC , color='goldenrod',label='SC')
ax3.set_xlabel('Age (Gyr)')
ax3.set_ylabel(r'rad_he_core')
ax3.legend()

ax4.plot(Yc_OV,he_core_radius_OV, color='red')
ax4.plot(Yc_PC,he_core_radius_PC, color='darkgreen')
ax4.plot(Yc_maxOV,he_core_radius_maxOV, color='blue')
ax4.plot(Yc_SC ,he_core_radius_SC , color='goldenrod')
ax4.set_xlabel(r'$Y_c$')
ax4.set_ylabel(r'rad_he_core')
ax4.invert_xaxis()

plt.subplots_adjust(hspace = 0.4)

plt.show()
```
